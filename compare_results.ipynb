```python
# Step 0) Imports
import json
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

# Step 1) Define sources for all 3 models
# You can switch these paths if needed.
sources = {
    "Traditional": Path("traditional_results_latest.json"),
    "Semantic_v5": Path("semantic_results_all_bleu_latest.json"),
    "DeepSC": Path("deepsc_results_all_bleu_latest.json"),
}

# Step 2) Load JSONs and flatten bleu_by_snr into one dataframe
rows = []
for model_name, p in sources.items():
    if not p.exists():
        raise FileNotFoundError(f"Missing file: {p}")

    payload = json.loads(p.read_text(encoding="utf-8"))
    bleu_by_snr = payload.get("bleu_by_snr", {})

    for snr_str, scores in bleu_by_snr.items():
        rows.append({
            "model": model_name,
            "snr_db": int(snr_str),
            "bleu1": float(scores["bleu1"]),
            "bleu2": float(scores["bleu2"]),
            "bleu3": float(scores["bleu3"]),
            "bleu4": float(scores["bleu4"]),
        })

df = pd.DataFrame(rows).sort_values(["snr_db", "model"]).reset_index(drop=True)
print("Loaded comparison dataframe:")
display(df.head(10))

# Step 3) Create wide comparison tables (one per BLEU)
print("\nStep 3: BLEU tables across all SNRs")
for b in ["bleu1", "bleu2", "bleu3", "bleu4"]:
    pivot = df.pivot(index="snr_db", columns="model", values=b).sort_index()
    print(f"\n{b.upper()} comparison table:")
    display(pivot.round(4))

# Step 4) Plot BLEU-1/2/3/4 across all SNRs for all models
fig, axes = plt.subplots(2, 2, figsize=(13, 9), sharex=True)
axes = axes.flatten()
metrics = ["bleu1", "bleu2", "bleu3", "bleu4"]
titles = ["BLEU-1 vs SNR", "BLEU-2 vs SNR", "BLEU-3 vs SNR", "BLEU-4 vs SNR"]

for ax, metric, title in zip(axes, metrics, titles):
    for model_name in ["Traditional", "Semantic_v5", "DeepSC"]:
        sub = df[df["model"] == model_name].sort_values("snr_db")
        ax.plot(sub["snr_db"], sub[metric], marker="o", linewidth=2, label=model_name)

    ax.set_title(title)
    ax.set_xlabel("SNR (dB)")
    ax.set_ylabel(metric.upper())
    ax.grid(True, alpha=0.35)
    ax.legend()

plt.tight_layout()
plt.savefig("compare_all_models_bleu1234_vs_snr.png", dpi=170, bbox_inches="tight")
plt.show()
print("Saved: compare_all_models_bleu1234_vs_snr.png")

# Step 5) Quantitative gain tables (vs Traditional)
# Gain factor and absolute gain for each BLEU metric
trad = df[df["model"] == "Traditional"].set_index("snr_db")
sem  = df[df["model"] == "Semantic_v5"].set_index("snr_db")
deep = df[df["model"] == "DeepSC"].set_index("snr_db")

gain_rows = []
for snr in sorted(df["snr_db"].unique()):
    row = {"snr_db": snr}
    for b in ["bleu1", "bleu2", "bleu3", "bleu4"]:
        row[f"Semantic_vs_Traditional_{b}_gain_x"] = sem.loc[snr, b] / trad.loc[snr, b]
        row[f"DeepSC_vs_Traditional_{b}_gain_x"]   = deep.loc[snr, b] / trad.loc[snr, b]
        row[f"DeepSC_vs_Semantic_{b}_delta"]       = deep.loc[snr, b] - sem.loc[snr, b]
    gain_rows.append(row)

gain_df = pd.DataFrame(gain_rows).sort_values("snr_db")
print("\nStep 5: Gain table")
display(gain_df.round(4))

# Step 6) Winner per SNR per BLEU metric
winner_rows = []
for snr in sorted(df["snr_db"].unique()):
    row = {"snr_db": snr}
    for b in ["bleu1", "bleu2", "bleu3", "bleu4"]:
        sub = df[df["snr_db"] == snr][["model", b]].sort_values(b, ascending=False)
        row[f"best_{b}"] = sub.iloc[0]["model"]
        row[f"best_{b}_value"] = float(sub.iloc[0][b])
    winner_rows.append(row)

winner_df = pd.DataFrame(winner_rows)
print("\nStep 6: Best model at each SNR")
display(winner_df.round(4))

# Step 7) Save final merged outputs for report
df.to_csv("compare_all_models_bleu1234_long.csv", index=False)
gain_df.to_csv("compare_all_models_bleu1234_gains.csv", index=False)
winner_df.to_csv("compare_all_models_bleu1234_winners.csv", index=False)

print("\nSaved:")
print(" - compare_all_models_bleu1234_long.csv")
print(" - compare_all_models_bleu1234_gains.csv")
print(" - compare_all_models_bleu1234_winners.csv")
```
